```python
# if-else
# case-when in Spark SQL | when-otherwise in PySpark
from pyspark.sql.functions import *
from pyspark.sql.types import *

mngr_df.withColumn("high_salary",when(col("salary") > 60000, "Yes")
                   .when(col("salary") < 60000, "No")
                   .otherwise("Equal")).show()
```

```python
mngr_df.withColumn("salary_adjusted", when(col("salary").isNull(), lit(50000))
                   .otherwise(col("salary")))\
    .withColumn("salary_category", when((col("salary_adjusted") > 60000) & (col("salary_adjusted") < 80000), "Medium")
        .when(col("salary_adjusted") > 80000, "High")
        .when(col("salary_adjusted") < 60000, "Low")
        .otherwise("Equal")).show()
```

```python
# find out the performance of sales based on last 3 months average
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql.window import Window

windowSpec = Window.partitionBy('product_id').orderBy('sales_date').rowsBetween(-2, 0)
windowSpec1 = Window.partitionBy('product_id').orderBy('sales_date')

df = product_df.withColumn('running_sales_sum', sum('sales').over(windowSpec))

# filtering out the first 2 records for each product
filtered_df = df.withColumn('rw_number', row_number().over(windowSpec1))\
                .filter(col('rw_number') > 2)\
                .drop('rw_number')

avg_df = filtered_df.withColumn('avg_sales', round(col('running_sales_sum')/3, 2))

avg_df.show()

# filtering out the last 3 months best performance sales of product
best_df = avg_df.filter(col('avg_sales') <= col('sales')).drop('running_sales_sum')

best_df.show()
```
